## Numpy copy vs view

```python
# x1=
x1 = np.array([10, 20, 30, 40, 50])
```

```python
# assign x2 = x1
x2 = x1
```

```python
#print x1 and x2
print(f"x1: {x1}")
print(f"x2: {x2}")
```

Ok now you have seen that both of them are same

```python
# change 1st element of x2 as 10
x2[0] = 10
```

```python
#Again print x1 and x2
print(f"x1 after x2 change: {x1}")
print(f"x2 after x2 change: {x2}")
```

```python
# Check memory share between x1 and x2
print(np.shares_memory(x1, x2))
```

Hey It's True they both share memory

Shall we try **view()** function also likwise.

```python
# Create a view of x1 and store it in x3.
x3 = x1.view()
print(f"x3 (view of x1): {x3}")
```

```python
# Again check memory share between x1 and x3
print(np.shares_memory(x1, x3))
```

Woh! simple assignment is similar to view. That means
The view does not own the data and any changes made to the view will affect the original array, and any changes made to the original array will affect the view.

Don't agree? ok lets change x3 and see if original array i.e. x1 also changes

```python
#Change 1st element of x3=100
x3[0] = 100
```

```python
#print x1 and x3 to check if changes reflected in both
print(f"x1 after x3 change: {x1}")
print(f"x3 after x3 change: {x3}")
```

Now its proved.

Lets see how **Copy()** function works

```python
# Now create an array x4 which is copy of x1
x4 = x1.copy()
print(f"x4 (copy of x1): {x4}")
```

```python
# Change the last element of x4 as 900
x4[-1] = 900
print(f"x4 after modification: {x4}")
```

```python
# print both x1 and x4 to check if changes reflected in both
print(f"x1 after x4 modification: {x1}")
print(f"x4 after x4 modification: {x4}")
```

**hstack vs vstack function**


Stacking is same as concatenation, the only difference is that stacking is done along a new axis.

NumPy provides a helper function:

1. hstack() to stack along rows.
2. vstack()  to stack along columns

```python
# stack x1 and x4 along columns.
# For hstack, arrays must have the same number of rows or be 1D.
# Let's use x1 and x4 directly as they are 1D.
h_stacked = np.hstack((x1, x4))
print("h_stacked:", h_stacked)
```

```python
#stack x1 and x4 along rows
# For vstack, arrays must have the same shape along the axis being stacked.
# If they are 1D, vstack treats them as rows and stacks them into a 2D array.
v_stacked = np.vstack((x1, x4))
print("v_stacked:\n", v_stacked)
```

We hope now you saw the difference between them.

Fun fact! you can even use concatenate() function to join 2 arrays along with the axis. If axis is not explicitly passed, it is taken as 0 ie. along column

Lets try this function as well

```python
# arr1 =
arr1 = np.array([1, 2, 3])
# arr2 =
arr2 = np.array([4, 5, 6])

##join arr1 and arr2 along rows using concatenate() function
concatenated_rows = np.concatenate((arr1, arr2), axis=0) # axis=0 for 1D arrays is simply concatenation
print("Concatenated along rows (axis=0):", concatenated_rows)
```

```python
##join arr1 and arr2 along columns using concatenate() function
# To concatenate along columns (axis=1), arrays need to be 2D
arr1_2d = arr1.reshape(1, -1)
arr2_2d = arr2.reshape(1, -1)
concatenated_cols = np.concatenate((arr1_2d, arr2_2d), axis=1)
print("Concatenated along columns (axis=1):\n", concatenated_cols)
```

## Adding, Insert and delete Numpy array

You can also add 2 arrays using append() function also. This function appends values to end of array

Lets see how

```python
# append arr2 to arr1
appended_array = np.append(arr1, arr2)
print("Appended array:", appended_array)
```

Lets use insert() function which Inserts values into array before specified index value

```python
# Inserts values into array x1 before index 4 with elements of x4
inserted_array = np.insert(x1, 4, x4)
print("Array after insertion:", inserted_array)
```
