## Applying functions to columns and creating new columns

We need to create another column in d3, which contains  a boolean value for each city to indicate whether it's a union territory or not.
- HINT: Chandigarh, Pondicherry and Delhi are only 3 union territories here.


```python
def is_ut(x):
    if x in ['Chandigarh', 'Pondicherry', 'Delhi']:
        return True
    else:
        return False

d3['is_ut'] = d3['city'].apply(is_ut)
```

```python
print(d3)
```

The temperatures mentioned in ‘Temperature’ column are mentioned in Celsius, we need another column which contains the same in Fahrenheit.

HINT -
- Define a function c_to_f which takes input temp in celsius and returns a value with temperature in Fahrenheit.
- To check: c_to_f(10) should return 50.


```python
def c_to_f(celsius):
    return (celsius * 9/5) + 32
```

```python
print(f"10 Celsius in Fahrenheit is: {c_to_f(10)}")
```

```python
d3['temp_farenheit'] = d3['Temperature'].apply(c_to_f)
print(d3)
```

## Groupby

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'C', 'C'],
    'Value': [10, 20, 15, 25, 12, 30, 18]
}
df_groupby = pd.DataFrame(data)
print(df_groupby)
```

```python
print(df_groupby.groupby('Category')['Value'].sum())
```

```python
print(df_groupby.groupby('Category')['Value'].count())
```

```python
print(df_groupby.groupby('Category')['Value'].agg(['min', 'max']))
```

```python
data_two_cols = {
    'Category1': ['X', 'Y', 'X', 'Y', 'Z', 'Z', 'X'],
    'Category2': ['P', 'Q', 'P', 'Q', 'R', 'R', 'P'],
    'Score': [100, 200, 150, 250, 120, 300, 180]
}
df_groupby_two_cols = pd.DataFrame(data_two_cols)
print(df_groupby_two_cols.groupby(['Category1', 'Category2'])['Score'].mean())
```

```python
print(df_groupby_two_cols.groupby(['Category1', 'Category2'])['Score'].agg(['min', 'max']))
```
