**TASK - 2**

```python
plt.figure(figsize=(10,5))

sns.scatterplot(x='age', y='salary', data = df, color = 'blue', edgecolor='linen', alpha = 0.5)
plt.xlabel('Employee Age')
plt.ylabel('Employee Salary')
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.scatterplot(x='healthy_eating', y='active_lifestyle', data = df, color = 'red', edgecolor='linen', alpha = 0.5)
plt.xlabel('Healthy Eating')
plt.ylabel('Active Lifestyle ')
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.scatterplot(x='healthy_eating', y='salary', data = df, color = 'k', edgecolor='linen', alpha = 0.5)
plt.xlabel('Healthy Eating')
plt.ylabel('Salary')
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(x='groups', data = df, edgecolor='linen', alpha = 0.7)
plt.xlabel('Groups')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(x='salary', data = df, edgecolor='linen', alpha = 0.7, bins = 10, color = 'orange')
plt.xlabel('Salary')
plt.ylabel('Count')
plt.show()
```
