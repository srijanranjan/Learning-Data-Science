```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

**TASK - 1**

```python
df = pd.read_csv('Wine Quality Dataset.csv')
```

**TASK - 2**


```python
plt.figure(figsize = (11,6))

sns.histplot(data = df, x = 'fixed acidity', color = 'orange', alpha = 0.5, bins = 5)

plt.title('Histogram For Fixed Acidity')
plt.xlabel('Fixed Acidity')
plt.ylabel('Count')
plt.show()
```

```python
print(df['fixed acidity'].mean())
```

```python
print(df['fixed acidity'].median())
```

```python
plt.figure(figsize=(11,6))

sns.histplot(data = df, x = 'fixed acidity', color = 'orange',alpha = 0.5, edgecolor= 'linen', bins = 5)
plt.xlabel('Fixed Acidity')
plt.ylabel('Count')
plt.vlines(df['fixed acidity'].mean(), ymin = 0, ymax = 4000, colors = 'blue', label = 'Mean')
plt.vlines(df['fixed acidity'].median(), ymin= 0, ymax = 4000, colors = 'red', label = 'Median')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(11,6))

sns.histplot(data = df, x = 'volatile acidity', color = 'green', alpha = 0.5, bins = 5)

plt.title('Histogram For Volatile Acidity')
plt.xlabel('Volatile Acidity')
plt.ylabel('Count')
plt.show()
```

```python
df['volatile acidity'].skew()
```

```python
plt.figure(figsize=(11,6))

sns.distplot(df['volatile acidity'], color = 'blue')

plt.title('Distplot For Volatile Acidity')
plt.xlabel('Volatile Acidity')
plt.ylabel('Density')
plt.show()
```
